### #1.2

Crea una función similar a la que se pide en el ejercicio 6 de la segunda tanda de ejercicios del workout. Es decir, una función para pintar diagramas de dispersion que tenga seis argumentos, pero ojo al cambio en el quinto:
1. Un dataframe
2. Una columna numérica del dataframe
3. Una columna numérica del dataframe diferente a la anterior.
4. Un columna categórica (Opcional, valor por defecto None)
5. Un valor para el tamaño de los puntos o una cadena, pero que puede ser una cadena de caracteres, en cuyo caso se debe interpretar como una de las columnas del dataframe cuyos valores deben controlar el tamaño
6. Un sexto argumento "escala", que permita manipular la escala del tamaño en el caso de que el quinto argumento sea una cadena. Este argumento multiplicará a los valores de la columna indicada en el parámetro anterior (así si quiero dividir por 10000, este sexto argumento tendrá que valer 1/10000)

La función debe pintar el diagrama de dispersión de la columna numérica 1, en el eje X, frente a la columna numérica 2, en el eje Y, usando un código de colores en función de la columna categórica y con el tamaño de los puntos según el argumento correspondiente. 

Si no se diera una columna categórica como argumento, el color no debe pasarse como parámetro a la función de Seaborn o Matplotlib que uses.

EXTRA OPCIONAL: Añade un argumento que controle si la función pinta una leyenda para colores y tamaños o un colorbar y una leyenda para los tamaños.

```python


def cardinalidad(df_in: pd.DataFrame, umbral_categoria: int, umbral_continua: float) -> pd.DataFrame:
    resultados = []

    n_filas = len(df_in)

    for col in df_in.columns:
        valores_unicos = df_in[col].nunique(dropna=False)
        porcentaje = valores_unicos / n_filas

        if valores_unicos == 2:
            tipo = "Binaria"
        elif valores_unicos < umbral_categoria:
            tipo = "Categórica"
        else:
            if porcentaje >= umbral_continua:
                tipo = "Numerica Continua"
            else:
                tipo = "Numerica Discreta"

        resultados.append({
            "columna": col,
            "cardinalidad": valores_unicos,
            "porcentaje_cardinalidad": porcentaje,
            "tipo": tipo
        })

    return pd.DataFrame(resultados)
```

```python
cardinalidad(df,0.1,0.1)
```

```python

def scatter_plot(df, col_x, col_y, col_cat=None, size=50, escala=1):
    # Calcular el tamaño de los puntos
    if isinstance(size, str):
        sizes = df[size] * escala
    else:
        sizes = size
    
    fig, ax = plt.subplots(figsize=(10, 6))
    
    if col_cat is not None:
        # Con columna categórica → color por categoría
        categorias = df[col_cat].unique()
        for cat in categorias:
            mask = df[col_cat] == cat
            ax.scatter(
                df.loc[mask, col_x],
                df.loc[mask, col_y],
                s=sizes[mask] if isinstance(size, str) else sizes,
                label=cat,
                alpha=0.6
            )
        ax.legend(title=col_cat)
    else:
        # Sin columna categórica → sin color
        ax.scatter(df[col_x], df[col_y], s=sizes, alpha=0.6)
    
    ax.set_xlabel(col_x)
    ax.set_ylabel(col_y)
    ax.set_title(f"{col_x} vs {col_y}")
    plt.tight_layout()
    plt.show()
```

```python
scatter_plot(df,col_x="city", col_y="population_total", size="area_total_km2", escala=1/10)
```

```python
df_titanic=pd.read_csv("./data/titanic.csv")
display(df_titanic)
```

```python

tasa = df_titanic['survived'].value_counts(normalize=True) * 100

fig, ax = plt.subplots(figsize=(6, 4))

sns.countplot(data=df_titanic, x='survived', palette=['#e74c3c', '#2ecc71'], ax=ax)

ax.set_xticklabels(['No sobrevivió', 'Sobrevivió'])
ax.set_title("Tasa de mortalidad")
ax.set_ylabel('Pasajeros')
plt.tight_layout()
plt.show()
```

```python
df_titanic["grupo"] = df_titanic["age"].apply(lambda x: "Niño (<18)" if x < 18 else "Adulto")

surv_sexo = df_titanic.groupby("who")["survived"].mean() * 100
surv_grupo = df_titanic.groupby("grupo")["survived"].mean() * 100

fig, axes = plt.subplots(1, 2, figsize=(12, 5))

axes[0].bar(surv_sexo.index, surv_sexo.values, color=["#3498db", "#e74c3c"])
axes[0].set_title("Supervivencia por sexo")
axes[0].set_ylabel("% Supervivencia")
for i, v in enumerate(surv_sexo.values):
    axes[0].text(i, v + 1, f"{v:.1f}%", ha="center", fontweight="bold")

axes[1].bar(surv_grupo.index, surv_grupo.values, color=["#f39c12", "#9b59b6"])
axes[1].set_title("Supervivencia por grupo de edad")
axes[1].set_ylabel("% Supervivencia")
for i, v in enumerate(surv_grupo.values):
    axes[1].text(i, v + 1, f"{v:.1f}%", ha="center", fontweight="bold")

plt.tight_layout()
plt.show()
```

```python
surv_sexo_clase = df_titanic.groupby(["pclass", "who"])["survived"].mean().unstack() * 100
surv_ninos_clase = df_titanic[df_titanic["grupo"] == "Niño (<18)"].groupby("pclass")["survived"].mean() * 100

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

surv_sexo_clase.plot(kind="bar", ax=axes[0], color=["#e74c3c", "#3498db", "#f1c40f"], rot=0)
axes[0].set_title("Supervivencia por sexo y clase")
axes[0].set_xlabel("Clase")
axes[0].set_ylabel("% Supervivencia")
axes[0].set_xticklabels(["1ª Clase", "2ª Clase", "3ª Clase"])
axes[0].legend(title="Sexo")

axes[1].bar(["1ª Clase", "2ª Clase", "3ª Clase"], 
             surv_ninos_clase.values, 
             color=["#f1c40f", "#95a5a6", "#cd6155"])
axes[1].set_title("Supervivencia de niños por clase")
axes[1].set_ylabel("% Supervivencia")
for i, v in enumerate(surv_ninos_clase.values):
    axes[1].text(i, v + 1, f"{v:.1f}%", ha="center", fontweight="bold")

plt.tight_layout()
plt.show()
```

```python
surv_embarque = df_titanic.groupby("embarked")["survived"].mean() * 100
clase_embarque = df_titanic.groupby(["embarked", "pclass"]).size().unstack(fill_value=0)
clase_embarque_pct = clase_embarque.div(clase_embarque.sum(axis=1), axis=0) * 100

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

axes[0].bar(surv_embarque.index, surv_embarque.values, color=["#1abc9c", "#e67e22", "#8e44ad"])
axes[0].set_title("Supervivencia por puerto de embarque")
axes[0].set_ylabel("% Supervivencia")
axes[0].set_xticklabels(["Cherbourg (C)", "Queenstown (Q)", "Southampton (S)"])
for i, v in enumerate(surv_embarque.values):
    axes[0].text(i, v + 1, f"{v:.1f}%", ha="center", fontweight="bold")

clase_embarque_pct.plot(kind="bar", ax=axes[1], color=["#f1c40f", "#95a5a6", "#cd6155"], rot=0)
axes[1].set_title("Distribución de clases por puerto")
axes[1].set_ylabel("% Pasajeros")
axes[1].set_xticklabels(["Cherbourg", "Queenstown", "Southampton"])
axes[1].legend(title="Clase", labels=["1ª", "2ª", "3ª"])

plt.tight_layout()
plt.show()
```
